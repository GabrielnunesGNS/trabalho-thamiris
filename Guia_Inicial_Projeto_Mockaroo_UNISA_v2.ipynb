## 14. Medição de tempo

Use `perf_counter()`. Nunca tire conclusão de uma única execução.

````python
from time import perf_counter

def medir_uma_execucao(funcao, *args):
    inicio = perf_counter()
    resultado = funcao(*args)
    fim = perf_counter()
    return resultado, fim - inicio
````

````python
def repetir_medicao(funcao, repeticoes, *args):
    tempos = []
    for _ in range(repeticoes):
        inicio = perf_counter()
        funcao(*args)
        fim = perf_counter()
        tempos.append(fim - inicio)
    return tempos
````

## 15. Média, mediana e desvio padrão

````python
import statistics

def resumir_tempos(tempos):
    return {
        "media": statistics.mean(tempos),
        "mediana": statistics.median(tempos),
        "desvio_padrao": statistics.stdev(tempos) if len(tempos) > 1 else 0
    }

tempos_exemplo = [0.10, 0.12, 0.11, 0.09, 0.13]
resumir_tempos(tempos_exemplo)
````
